# RWB — Ridge baseline

Этот ноутбук запускает цельный скрипт обучения из корня проекта. Он заменяет разрозненные ячейки, в которых ранее возникали `NameError` из-за порядка выполнения.
Скрипт создаст `submissions/ridge_baseline_submission.csv`, сохранит модель в `models/` и отчёты в `reports/`.

```python
from pathlib import Path
import subprocess
import sys

candidates = [
    Path.cwd() / 'train_ridge_baseline.py',
    Path.cwd().parent / 'train_ridge_baseline.py',
]
script_path = next((p.resolve() for p in candidates if p.is_file()), None)

if script_path is None:
    raise FileNotFoundError(
        'Не найден train_ridge_baseline.py. Положи его в корень RWB_hack.'
    )

print('Python:', sys.executable)
print('Script:', script_path)
result = subprocess.run([sys.executable, str(script_path)], cwd=script_path.parent)
if result.returncode != 0:
    raise RuntimeError(f'Скрипт завершился с кодом {result.returncode}')
```
